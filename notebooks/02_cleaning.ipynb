# Conclusiones de Cleaning

## Principales hallazgos

* No se encontraron valores nulos en ninguna de las variables del dataset.

* Se identificaron **1.081 registros duplicados**, de los cuales **1.062** correspondían a transacciones legítimas y **19** a transacciones fraudulentas.

* Los registros duplicados eran duplicados exactos, es decir, presentaban los mismos valores en todas las variables.

* El dataset no contiene una columna identificadora que justifique la conservación de registros idénticos.

## Decisiones de limpieza

* No fue necesario aplicar técnicas de imputación ni eliminar registros por valores faltantes debido a la ausencia de valores nulos.

* Se eliminaron los **1.081 registros duplicados** identificados durante el análisis.

* No fue necesario realizar transformaciones adicionales de limpieza sobre las variables del dataset.

## Resultado final

* El número de registros pasó de **284.807** a **283.726** tras la eliminación de duplicados.

* No se eliminaron columnas y el dataset conservó sus **31 variables originales**.

* Tras la limpieza, no se detectaron valores nulos ni registros duplicados.
