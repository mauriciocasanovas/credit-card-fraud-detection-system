Dado que se trata de un problema de clasificación altamente desbalanceado, la Accuracy no será la métrica principal de evaluación. En su lugar, se utilizarán métricas más adecuadas para la detección de fraude, como Recall, Precision, F1-score, ROC-AUC y PR-AUC.

## Recall

Recall mide la capacidad de un modelo para detectar correctamente las transacciones fraudulentas.

Esta métrica responde a la siguiente pregunta:

> De todos los fraudes reales, ¿cuántos logró identificar el modelo?

En problemas de detección de fraude, Recall suele ser una de las métricas más importantes, ya que un bajo Recall implica dejar pasar transacciones fraudulentas sin detectar.

```python
import pandas as pd
```

```python
recall_df = pd.DataFrame({
    "Modelo": ["Logistic Regression", "Random Forest", "XGBoost"],
    "Recall": [0.58, 0.73, 0.69]
})

recall_df
```

### Interpretación

Los resultados muestran diferencias claras entre los modelos evaluados.

Random Forest obtuvo el Recall más alto (73%), logrando detectar la mayor proporción de transacciones fraudulentas presentes en el conjunto de prueba.

XGBoost alcanzó un Recall del 69%, mostrando también una buena capacidad de detección, aunque ligeramente inferior a Random Forest.

Por su parte, Logistic Regression obtuvo el menor Recall (58%), dejando sin detectar una mayor cantidad de fraudes.

Dado que el objetivo principal del proyecto es identificar transacciones fraudulentas, Random Forest presenta el mejor desempeño según esta métrica.

## Precision

Precision mide la proporción de transacciones identificadas como fraude que realmente corresponden a fraudes.

Esta métrica responde a la siguiente pregunta:

> De todas las transacciones que el modelo clasificó como fraude, ¿cuántas eran realmente fraudulentas?

Una Precision elevada indica que el modelo genera pocas falsas alarmas, evitando clasificar incorrectamente transacciones legítimas como fraudulentas.

```python
precision_df = pd.DataFrame({
    "Modelo": ["Logistic Regression", "Random Forest", "XGBoost"],
    "Precision": [0.85, 0.97, 0.75]
})

precision_df
```

### Interpretación

Random Forest obtuvo la Precision más alta (97%), lo que indica que la gran mayoría de las transacciones clasificadas como fraude eran realmente fraudulentas.

Logistic Regression alcanzó una Precision del 85%, mostrando también una baja proporción de falsas alarmas.

Por su parte, XGBoost obtuvo una Precision del 75%, siendo el modelo con peor desempeño en esta métrica debido a una mayor cantidad de falsos positivos.

Dado que una Precision elevada reduce el número de transacciones legítimas clasificadas erróneamente como fraude, Random Forest vuelve a posicionarse como el modelo con mejor rendimiento.

## F1-Score

El F1-score combina Precision y Recall en una única métrica, proporcionando una medida equilibrada del rendimiento del modelo.

Esta métrica resulta especialmente útil en problemas de clasificación desbalanceada, donde es importante detectar la mayor cantidad posible de fraudes sin generar un número excesivo de falsas alarmas.

```python
f1_df = pd.DataFrame({
    "Modelo": ["Logistic Regression", "Random Forest", "XGBoost"],
    "F1-Score": [0.69, 0.83, 0.72]
})

f1_df
```

### Interpretación

Random Forest obtuvo el F1-score más alto (83%), reflejando el mejor equilibrio entre Precision y Recall.

XGBoost alcanzó un F1-score del 72%, mostrando un rendimiento intermedio entre los modelos evaluados.

Logistic Regression obtuvo el F1-score más bajo (69%), debido principalmente a su menor capacidad para detectar fraudes.

Considerando simultáneamente la capacidad de detección de fraude y la reducción de falsas alarmas, Random Forest presenta nuevamente el mejor desempeño global.

## PR-AUC

PR-AUC (Precision-Recall Area Under the Curve) evalúa el equilibrio entre Precision y Recall para distintos umbrales de decisión.

Esta métrica resulta especialmente útil en problemas con clases altamente desbalanceadas, como la detección de fraude, donde la clase positiva representa una proporción muy pequeña del total de observaciones.

A diferencia de ROC-AUC, PR-AUC se centra específicamente en el rendimiento sobre la clase positiva, proporcionando una evaluación más exigente de la capacidad del modelo para detectar fraudes.

```python
precision_lr, recall_lr, _ = precision_recall_curve(y_test, y_prob_lr)
precision_rf, recall_rf, _ = precision_recall_curve(y_test, y_prob_rf)
precision_xgb, recall_xgb, _ = precision_recall_curve(y_test, y_prob_xgb)

pr_auc_df = pd.DataFrame({
    "Modelo": ["Logistic Regression", "Random Forest", "XGBoost"],
    "PR-AUC": [
        auc(recall_lr, precision_lr),
        auc(recall_rf, precision_rf),
        auc(recall_xgb, precision_xgb)
    ]
})

pr_auc_df
```

### Interpretación

Random Forest obtuvo el mayor valor de PR-AUC (0.797), mostrando el mejor equilibrio entre Precision y Recall a distintos umbrales de decisión.

XGBoost alcanzó un valor de 0.694, mientras que Logistic Regression obtuvo 0.690, presentando resultados muy similares entre sí.

Dado que el problema de detección de fraude presenta una fuerte desproporción entre clases, PR-AUC constituye una métrica especialmente relevante. En este contexto, Random Forest vuelve a destacar como el modelo con mejor desempeño global.
